# PHASE 5 - Data Cleaning

```python
# make backup copies of all files 
customers_clean = customers.copy()
accounts_clean = accounts.copy()
cards_clean = cards.copy()
complaints_clean = complaints.copy()
fraud_cases_clean = fraud_cases.copy()
loans_clean = loans.copy()
transactions_clean = transactions.copy()

print("Backup copies created successfully!")
```

```python
# Customer
customers_clean["Occupation"] = (
    customers_clean["Occupation"]
    .str.strip()
    .str.title()
)
```

```python
# Loan
loans_clean["Loan_Type"] = (
    loans_clean["Loan_Type"]
    .str.strip()
    .str.title()
)

loans_clean["Status"] = (
    loans_clean["Status"]
    .str.strip()
    .str.title()
)
```

```python
# Transactions:For transaction types, we'll use a controlled mapping instead.
# Replace the previous transaction Type operation with this:
transactions_clean["Type"] = (
    transactions_clean["Type"]
    .str.strip()
    .str.upper()
)

transaction_type_map = {
    "UPI": "UPI",
    "TRANSFER": "Transfer",
    "ATM": "ATM",
    "ONLINE": "Online",
    "BILL PAYMENT": "Bill Payment",
    "POS": "POS"
}

transactions_clean["Type"] = transactions_clean["Type"].map(transaction_type_map)

transactions_clean["Debit_Credit"] = (
    transactions_clean["Debit_Credit"]
    .str.strip()
    .str.title()
)
```

```python
print("Customer Occupations:")
print(customers_clean["Occupation"].unique())

print("\nLoan Types:")
print(loans_clean["Loan_Type"].unique())

print("\nLoan Status:")
print(loans_clean["Status"].unique())

print("\nTransaction Types:")
print(transactions_clean["Type"].unique())

print("\nDebit/Credit:")
print(transactions_clean["Debit_Credit"].unique())
```

```python
# Handle Missig Values
# Customer name
customers_clean["Customer_Name"] = (
    customers_clean["Customer_Name"]
    .fillna("Unknown")
)

```

```python
#  Customer Age -> median
customers_clean["Age"] = (
    customers_clean["Age"]
    .fillna(customers_clean["Age"].median())
)
```

```python
# Customer-> Occupation
customers_clean["Occupation"] = (
    customers_clean["Occupation"]
    .fillna("Undefined")
)
```

```python
# Loans Missing Values
loans_clean = loans_clean.dropna(
    subset=["Customer_ID"]
)
```

```python
# Loan Type
loans_clean["Loan_Type"] = (
    loans_clean["Loan_Type"]
    .fillna("Undefined")
)
```

```python
# loan Amount
loans_clean["Loan_Amount"] = (
    loans_clean["Loan_Amount"]
    .fillna(loans_clean["Loan_Amount"].median())
)
```

```python
# Transactions Missing Values
transactions_clean = transactions_clean.dropna(
    subset=["Account_ID", "Transaction_Date", "Type"]
)
```

```python
accounts_clean["Open_Date"] = pd.to_datetime(
    accounts_clean["Open_Date"],
    errors="coerce"
)

complaints_clean["Complaint_Date"] = pd.to_datetime(
    complaints_clean["Complaint_Date"],
    errors="coerce"
)

transactions_clean["Transaction_Date"] = pd.to_datetime(
    transactions_clean["Transaction_Date"],
    errors="coerce"
)
```

# PHASE 8 - Numpy

## 8A Transaction Business Features

```python
#Credit → positive amount
#Debit → negative amount
transactions_clean["Net_Amount"] = np.where(
    transactions_clean["Debit_Credit"] == "Credit",
    transactions_clean["Amount"],
    -transactions_clean["Amount"]
)
```

```python
transactions_clean[
    [
        "Amount",
        "Debit_Credit",
        "Net_Amount"
    ]
].head(10)
```

```python
# Loan Interest Amount
loans_clean["Estimated_Interest"] = (
    loans_clean["Loan_Amount"]
    * (loans_clean["Interest_Rate"] / 100)
    * (loans_clean["Tenure_Months"] / 12)
)
loans_clean[
    [
        "Loan_Amount",
        "Interest_Rate",
        "Tenure_Months",
        "Estimated_Interest"
    ]
].head(10)
```

```python
# Customer Financial Features
#calculate how many accounts each customer has and their total account balance.
customer_account_summary = (
    accounts_clean
    .groupby("Customer_ID")
    .agg(
        Total_Accounts=("Account_ID", "count"),
        Total_Balance=("Balance", "sum")
    )
    .reset_index()
)
customer_account_summary.head(10)
```

```python
#connect this summary back to the customer dataset:
customers_clean = customers_clean.merge(
    customer_account_summary,
    on="Customer_ID",
    how="left"
)
```

```python
# Fill customers who have no account:
customers_clean["Total_Accounts"] = customers_clean["Total_Accounts"].fillna(0)
customers_clean["Total_Balance"] = customers_clean["Total_Balance"].fillna(0)
```

# PHASE 10 -  Data Visualization - matplotlib

#### 10.1 Average Customer Balance by Occupation

```python
# Average Customer Balance by Occupation

occupation_balance = (
    customers_clean
    .groupby("Occupation")["Total_Balance"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

occupation_balance.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Average Customer Balance by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Total Balance")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

#### 10.2 Loan Default Rate by Loan Type

```python
# Which loan types have the highest proportion of defaulted loans?
# Loan Default Rate by Loan Type

loan_default_rate = (
    loans_clean
    .groupby("Loan_Type")
    .apply(
        lambda x: (x["Status"] == "Defaulted").mean() * 100,
        include_groups=False
    )
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

loan_default_rate.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Loan Default Rate by Loan Type")
plt.xlabel("Loan Type")
plt.ylabel("Default Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

#### 10.3 Monthly Transaction Volume Trend

```python
# Monthly Transaction Volume Trend  for after 3 months 

monthly_transactions = (
    transactions_clean
    .groupby(
        transactions_clean["Transaction_Date"].dt.to_period("M")
    )
    .size()
)

# Convert Period to timestamp
monthly_transactions.index = monthly_transactions.index.to_timestamp()

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_transactions.index,
    monthly_transactions.values,
    marker="o"
)

plt.title("Monthly Transaction Volume Trend")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")

# Show every 3rd month for a cleaner x-axis
plt.xticks(
    monthly_transactions.index[::3],
    monthly_transactions.index[::3].strftime("%b %Y"),
    rotation=45
)

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

#### 10.4 Annual Income vs Total Customer Balance

```python
# Annual Income vs Total Customer Balance by Occupation
# Does the income–balance relationship differ across customer occupation groups?
correlation = customers_clean[
    ["Annual_Income", "Total_Balance"]
].corr().loc["Annual_Income", "Total_Balance"]

print(
    "Correlation between Annual Income and Total Balance:",
    round(correlation, 3)
)

plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=customers_clean,
    x="Annual_Income",
    y="Total_Balance",
    hue="Occupation",
    alpha=0.5,
    s=45
)

plt.title("Annual Income vs Total Customer Balance by Occupation")
plt.xlabel("Annual Income")
plt.ylabel("Total Customer Balance")

plt.legend(
    title="Occupation",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

#### 10.5 Complaint Type vs Priority — Average Satisfaction

#### 10.11 Transaction Type Distribution by Channel

```python
# # How are different transaction types distributed across banking channels
# Transaction Type Distribution by Channel

transaction_channel = pd.crosstab(
    transactions_clean["Channel"],
    transactions_clean["Type"]
)

# Reorder transaction types
transaction_order = [
    "ATM",
    "Bill Payment",
    "Online",
    "POS",
    "Transfer",
    "UPI"
]

transaction_channel = transaction_channel[transaction_order]

# Blue + Green shades
colors = [
    "#8EC5F5",  # Light Blue
    "#7BC67B",  # Light Green
    "#2E8B57",  # Green
    "#5DADE2",  # Medium Blue
    "#2874C6",  # Blue
    "#145A9E"   # Dark Blue
]

ax = transaction_channel.plot(
    kind="bar",
    stacked=True,
    figsize=(13, 7),
    color=colors,
    edgecolor="white",
    linewidth=0.7
)

plt.title(
    "Transaction Type Distribution by Channel",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Transaction Channel",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Number of Transactions",
    fontsize=12,
    fontweight="bold"
)

plt.xticks(rotation=0)

# Add labels inside each stacked section
for i, channel in enumerate(transaction_channel.index):

    cumulative = 0

    for transaction_type in transaction_order:

        value = transaction_channel.loc[channel, transaction_type]

        if value > 0:
            ax.text(
                i,
                cumulative + value / 2,
                f"{value:,}",
                ha="center",
                va="center",
                fontsize=9,
                fontweight="bold",
                color="white"
            )

        cumulative += value

    # Total label above each bar
    ax.text(
        i,
        cumulative + 1500,
        f"{cumulative:,}",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

plt.legend(
    title="Transaction Type",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.grid(
    axis="y",
    alpha=0.2
)

plt.tight_layout()
plt.show()
```

#### 10.12 Customer Financial Profile by City — Cross-Dataset Analysis

```python
#How do customer financial profiles differ across cities when we consider balances, loan exposure, and credit score together
# 1. Aggregate account balance for each customer
customer_accounts = (
    accounts_clean
    .groupby("Customer_ID")
    .agg(
        Total_Balance=("Balance", "sum"),
        Total_Accounts=("Account_ID", "count")
    )
    .reset_index()
)

# 2. Aggregate loan exposure for each customer
customer_loans = (
    loans_clean
    .groupby("Customer_ID")
    .agg(
        Total_Loan_Amount=("Loan_Amount", "sum"),
        Total_Loans=("Loan_ID", "count")
    )
    .reset_index()
)

# 3. Combine Customers + Accounts + Loans
customer_financial = (
    customers_clean[
        ["Customer_ID", "City", "Credit_Score"]
    ]
    .merge(customer_accounts, on="Customer_ID", how="left")
    .merge(customer_loans, on="Customer_ID", how="left")
)

# Fill customers with no loans/accounts
customer_financial[
    ["Total_Balance", "Total_Accounts",
     "Total_Loan_Amount", "Total_Loans"]
] = customer_financial[
    ["Total_Balance", "Total_Accounts",
     "Total_Loan_Amount", "Total_Loans"]
].fillna(0)

# 4. Create city-level financial profile
city_profile = (
    customer_financial
    .groupby("City")
    .agg(
        Avg_Balance=("Total_Balance", "mean"),
        Avg_Loan_Exposure=("Total_Loan_Amount", "mean"),
        Avg_Credit_Score=("Credit_Score", "mean")
    )
)

# 5. Normalize each metric so they can be compared in one heatmap
city_profile_normalized = (
    city_profile - city_profile.mean()
) / city_profile.std()

print("City Financial Profile:")
display(city_profile.round(2))

# 6. Heatmap
plt.figure(figsize=(12, 7))

sns.heatmap(
    city_profile_normalized,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5,
    center=0
)

plt.title(
    "City Financial Profile — Balance, Loan Exposure & Credit Score",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Metric")
plt.ylabel("City")

plt.tight_layout()
plt.show()
```
